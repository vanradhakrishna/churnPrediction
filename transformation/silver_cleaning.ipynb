```python
from pyspark.sql.functions import col,sum
df=spark.table("ml_catalog.bronze.churn_raw")
#check nulls
null_counts=df.select([sum(col(c).isNull().cast("int")).alias(c) for c in df.columns]).collect()[0]
null_counts
df_clean = (
    df.drop("RowNumber", "CustomerId", "Surname")  # remove useless columns
      .dropDuplicates()                            # remove duplicates
      .na.drop()                                   # remove nulls
)
```

```python
from pyspark.sql.functions import col, sum

df_clean.select([
    sum(col(c).isNull().cast("int")).alias(c)
    for c in df_clean.columns
]).show()
```

```python
total = df_clean.count()
distinct = df_clean.dropDuplicates().count()

print("Total:", total)
print("Distinct:", distinct)
```

```python
df_clean.write.mode("overwrite").saveAsTable("ml_catalog.silver.churn")
spark.table("ml_catalog.silver.churn").display()
df_clean = spark.table("ml_catalog.silver.churn")
```
