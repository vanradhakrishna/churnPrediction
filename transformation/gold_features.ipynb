```python
from pyspark.sql.functions import col,when
# Step 1: Read from silver
df = spark.table("ml_catalog.silver.churn")
df_gold = (
            df    
    .withColumn("Gender_num", when(col("Gender") == "Male", 1).otherwise(0))

    # One-hot encode Geography
    .withColumn("Geo_France", when(col("Geography") == "France", 1).otherwise(0))
    .withColumn("Geo_Spain", when(col("Geography") == "Spain", 1).otherwise(0))
    .withColumn("Geo_Germany", when(col("Geography") == "Germany", 1).otherwise(0))
    
    .drop("Gender", "Geography")

    )
df_gold.write.mode("overwrite").saveAsTable("ml_catalog.gold.churn")
spark.table("ml_catalog.gold.churn").display()
df_gold = spark.table("ml_catalog.gold.churn")
```
